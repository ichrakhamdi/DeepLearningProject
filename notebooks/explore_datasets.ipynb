```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
train_path = "../data/concatenated/train.csv"
test_path = "../data/concatenated/test.csv"
```

```python
train_df = pd.read_csv(train_path)
test_df = pd.read_csv(test_path)
```

```python
# Bar plots of label_2 (Binary: ATTACK vs BENIGN)
sns.countplot(x='label_2', data=train_df)
plt.title("Train - label_2 Distribution")
plt.show()
```

```python
# Multiclass label_6
sns.countplot(x='label_6', data=train_df, order=train_df['label_6'].value_counts().index)
plt.title("Train - label_6 Distribution")
plt.xticks(rotation=45)
plt.show()
```

```python
feature = train_df.columns[0] 

sns.histplot(data=train_df, x=feature, hue='label_2', bins=30, kde=True)
plt.title(f"{feature} distribution by label_2")
plt.show()
```

```python
# for numeric features only
numeric_cols = train_df.select_dtypes(include=['float64', 'int64']).columns

plt.figure(figsize=(12, 8))
sns.heatmap(train_df[numeric_cols].corr(), cmap='coolwarm', annot=False)
plt.title("Feature Correlation (Train)")
plt.show()
```
